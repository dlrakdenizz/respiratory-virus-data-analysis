## Removing Unnecessary Columns

The `geography_type` column contains the same value
(`state`) for every observation. Since it does not
provide additional information for the current analysis,
it is removed from the working dataset.

The original raw dataset remains unchanged.

```python
# Create a working copy to preserve the original dataset
clean_df = df.copy()

# Remove the constant geography type column
clean_df = clean_df.drop(columns=["geography_type"])

# Verify the result
clean_df.info()
```

## Converting Date Columns

The `mmwr_week_start` and `mmwr_week_end` columns
are stored as strings in the raw dataset.

These columns will be converted to datetime objects
to enable date-based preprocessing and feature extraction.

```python
# Convert date columns from string to datetime
clean_df["mmwr_week_start"] = pd.to_datetime(
    clean_df["mmwr_week_start"]
)

clean_df["mmwr_week_end"] = pd.to_datetime(
    clean_df["mmwr_week_end"]
)

# Verify the conversion
clean_df.info()
```

## Creating Date-Based Features

New features are extracted from the week start date
to provide additional temporal information.

The year and month are extracted directly from the date.
The MMWR week number is calculated using the CDC
epidemiological week convention.

```python
# Extract year and month from the week start date
clean_df["year"] = clean_df["mmwr_week_start"].dt.year
clean_df["month"] = clean_df["mmwr_week_start"].dt.month

# Inspect the new columns
clean_df[
    ["mmwr_week_start", "year", "month"]
].sample(10)
```

```python
from datetime import date, timedelta


def get_mmwr_week(value):
    """Return the CDC MMWR week number for a given date."""
    current_date = value.date()

    # CDC weeks start on Sunday.
    week_start = current_date - timedelta(
        days=(current_date.weekday() + 1) % 7
    )

    # MMWR week 1 contains January 4.
    jan4 = date(current_date.year, 1, 4)
    first_week_start = jan4 - timedelta(
        days=(jan4.weekday() + 1) % 7
    )

    if week_start < first_week_start:
        return get_mmwr_week(pd.Timestamp(
            current_date.year - 1, 12, 31
        ))

    next_jan4 = date(current_date.year + 1, 1, 4)
    next_first_week_start = next_jan4 - timedelta(
        days=(next_jan4.weekday() + 1) % 7
    )

    if week_start >= next_first_week_start:
        return 1

    return ((week_start - first_week_start).days // 7) + 1


clean_df["week"] = clean_df["mmwr_week_start"].apply(
    get_mmwr_week
)

clean_df[
    ["mmwr_week_start", "year", "month", "week"]
].sample(10)
```

## Examining Missing Values

The dataset contains missing observations in the
test_count, positive_count, and percent_pos columns.

Before handling these values, their distribution
across different pathogens will be examined.

```python
# Examine missing values by pathogen
missing_by_pathogen = clean_df.groupby("pathogen")[
    ["test_count", "positive_count", "percent_pos"]
].agg(lambda x: x.isna().sum())

missing_by_pathogen
```

```python
# Compare zero values and missing values
for column in ["test_count", "positive_count", "percent_pos"]:
    print(f"\n{column}")
    print("Missing:", clean_df[column].isna().sum())
    print("Zero:", (clean_df[column] == 0).sum())
```

### Missing Value Patterns

Missing values are examined together to identify
different patterns in laboratory test counts,
positive test counts, and positivity percentages.

Missing observations are not automatically replaced
with zero because some values are suppressed by CDC.

```python
# Examine combinations of missing values
columns = ["test_count", "positive_count", "percent_pos"]

missing_patterns = (
    clean_df[columns]
    .isna()
    .value_counts()
    .rename("record_count")
    .reset_index()
)

missing_patterns
```

```python
# Inspect records where no tests were performed
zero_test_records = clean_df[
    clean_df["test_count"] == 0
]

print("Records with zero tests:", len(zero_test_records))

print(
    zero_test_records[
        ["pathogen", "geography_value",
         "test_count", "positive_count", "percent_pos"]
    ].isna().sum()
)

zero_test_records[
    ["pathogen", "geography_value",
     "test_count", "positive_count", "percent_pos"]
].sample(min(10, len(zero_test_records)))
```

### Examining Suppressed Positive Test Counts

Records with available test counts but missing positive
test counts are examined to understand possible
suppression patterns in the CDC dataset.

```python
# Records with available test counts and missing positive counts
missing_positive = clean_df[
    clean_df["test_count"].notna()
    & clean_df["positive_count"].isna()
]

print("Records:", len(missing_positive))

# Inspect the distribution of test counts
print(missing_positive["test_count"].describe())

# View random examples
missing_positive[
    ["pathogen", "geography_value",
     "test_count", "positive_count", "percent_pos"]
].sample(15)
```

```python
# Verify the relationship between test counts and positivity

# Check records where test count is zero
zero_tests = clean_df[clean_df["test_count"] == 0]

print("Zero test records:", len(zero_tests))
print(
    "Zero test records with missing positivity:",
    zero_tests["percent_pos"].isna().sum()
)

# Check records with complete numeric information
complete_records = clean_df.dropna(
    subset=["test_count", "positive_count", "percent_pos"]
)

print("Records with complete numeric data:", len(complete_records))
print("Total records preserved:", len(clean_df))
```
