## Removing Unnecessary Columns

The `geography_type` column contains the same value
(`state`) for every observation. Since it does not
provide additional information for the current analysis,
it is removed from the working dataset.

The original raw dataset remains unchanged.

```python
# Create a working copy to preserve the original dataset
clean_df = df.copy()

# Remove the constant geography type column
clean_df = clean_df.drop(columns=["geography_type"])

# Verify the result
clean_df.info()
```

## Converting Date Columns

The `mmwr_week_start` and `mmwr_week_end` columns
are stored as strings in the raw dataset.

These columns will be converted to datetime objects
to enable date-based preprocessing and feature extraction.

```python
# Convert date columns from string to datetime
clean_df["mmwr_week_start"] = pd.to_datetime(
    clean_df["mmwr_week_start"]
)

clean_df["mmwr_week_end"] = pd.to_datetime(
    clean_df["mmwr_week_end"]
)

# Verify the conversion
clean_df.info()
```
